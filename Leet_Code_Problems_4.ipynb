2385. Amount of Time for Binary Tree to Be Infected

```python
class Solution(object):
    def amountOfTime(self, root, start):
        """
        :type root: Optional[TreeNode]
        :type start: int
        :rtype: int
        """

        from collections import deque

        # Build pointers to parents from children
        self.parent = {}
        self.start_node = None

        self.build_parents(root, None, start)

        level = [self.start_node]
        visited = set([self.start_node])
        minutes = -1

        while level:
            minutes += 1
            q = []
            for node in level:
                for neighbor in [self.parent[node], node.left, node.right]:
                    if neighbor and neighbor not in visited:
                        visited.add(neighbor)
                        q.append(neighbor)
            level = q

        return minutes

    def build_parents(self, node, par, start):
        if not node:
            return

        self.parent[node] = par

        if node.val == start:
            self.start_node = node

        self.build_parents(node.left, node, start)
        self.build_parents(node.right, node, start)
```

958. Check Completeness of a Binary Tree

```python
class Solution(object):
    def isCompleteTree(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: bool
        """

        level = [root]
        should_end = False
        while level:
            q = []
            for node in level:
                if should_end and (node.left or node.right):
                    return False
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
                if not node.left and node.right:
                    return False
                if node.left and not node.right:
                    should_end = True
                if not node.left and not node.right:
                    should_end = True
            level = q
        return True
```

993. Cousins in Binary Tree

```python
class Solution(object):
    def isCousins(self, root, x, y):
        """
        :type root: Optional[TreeNode]
        :type x: int
        :type y: int
        :rtype: bool
        """

        if not root:
            return False

        level = [root]

        while level:

            parent_x = None
            parent_y = None
            q = []

            for node in level:
                if node.left:
                    if node.left.val == x:
                        parent_x = node
                    if node.left.val == y:
                        parent_y = node
                    q.append(node.left)

                if node.right:
                    if node.right.val == x:
                        parent_x = node
                    if node.right.val == y:
                        parent_y = node
                    q.append(node.right)

                if parent_x and parent_y and parent_x != parent_y:
                    return True

                level = q

        return False
```

2641. Cousins in Binary Tree II

```python
class Solution(object):
    def replaceValueInTree(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: Optional[TreeNode]
        """

        if not root:
            return None

        root.val = 0

        level = [root]

        while level:
            q = []
            next_level_sum = 0
            for node in level:
                if node.left:
                    q.append(node.left)
                    next_level_sum += node.left.val
                if node.right:
                    q.append(node.right)
                    next_level_sum += node.right.val

            for node in level:
                if node.left and node.right:
                    node.left.val = next_level_sum - node.left.val - node.right.val
                    node.right.val = node.left.val

                elif node.right and not node.left:
                    node.right.val = next_level_sum - node.right.val

                elif node.left and not node.right:
                    node.left.val = next_level_sum - node.left.val

            level = q

        return root
```

623. Add One Row to Tree

```python
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def addOneRow(self, root, val, depth):
        """
        :type root: Optional[TreeNode]
        :type val: int
        :type depth: int
        :rtype: Optional[TreeNode]
        """

        if depth == 1:
            return TreeNode(val, root, None)

        level = [root]
        level_depth = 1

        while level:
            if level_depth == depth - 1:
                for node in level:
                    old_left = node.left
                    old_right = node.right

                    node.left = TreeNode(val)
                    node.right = TreeNode(val)

                    node.left.left = old_left
                    node.right.right = old_right

                break

            q = []
            for node in level:
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
            level = q
            level_depth += 1

        return root
```

1162. As Far from Land as Possible

```python
from collections import deque

class Solution(object):
    def maxDistance(self, grid):
        """
        :type grid: List[List[int]]
        :rtype: int
        """

        n = len(grid)
        q = deque()


        for i in range(n):
            for j in range(n):
                if grid[i][j] == 1:
                    q.append((i,j))

        if len(q) == 0 or len(q) == n*n:
            return -1

        directions = [(1,0), (-1,0), (0,1), (0,-1)]
        distance = -1

        while q:
            distance += 1
            for _ in range(len(q)):
                x, y = q.popleft()
                for dx, dy in directions:
                    nx, ny = x + dx, y + dy
                    if 0 <= nx < n and 0 <= ny < n and grid[nx][ny] == 0:
                        grid[nx][ny] = 1
                        q.append((nx, ny))

        return distance
```

1123. Lowest Common Ancestor of Deepest Leaves

```python
class Solution(object):
    def lcaDeepestLeaves(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: Optional[TreeNode]
        """

        depth, lca = self.dfs(root, 0)
        return lca

    def dfs(self, node, depth):

        if not node:
            return  depth - 1, None

        l_depth, l_lca = self.dfs(node.left, depth + 1)
        r_depth, r_lca = self.dfs(node.right, depth + 1)

        if l_depth == r_depth:
            return l_depth, node

        elif l_depth > r_depth:
            return l_depth, l_lca

        elif l_depth < r_depth:
            return r_depth, r_lca
```

1373. Maximum Sum BST in Binary Tree

```python
class Solution(object):
    def maxSumBST(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: int
        """

        self.max_sum = 0
        self.dfs(root)
        return self.max_sum

    def dfs(self, node):
        if not node:
            return True, float('inf'), float('-inf'), 0

        l_bst, l_min, l_max, l_sum = self.dfs(node.left)
        r_bst, r_min, r_max, r_sum = self.dfs(node.right)

        if l_bst and r_bst and l_max < node.val < r_min:

            curr_sum = node.val + l_sum + r_sum
            self.max_sum = max(self.max_sum, curr_sum)

            curr_min = min(l_min, node.val)
            curr_max = max(r_max, node.val)

            return True, curr_min, curr_max, curr_sum

        return False, 0, 0, 0
```

1110. Delete Nodes And Return Forest

```python
class Solution(object):
    def delNodes(self, root, to_delete):
        """
        :type root: TreeNode
        :type to_delete: List[int]
        :rtype: List[TreeNode]
        """

        to_delte = set(to_delete)
        self.forest = []
        root = self.dfs(root, to_delete)
        if root:
            self.forest.append(root)
        return self.forest

    def dfs(self, node, to_delete):
        if not node:
            return

        node.left = self.dfs(node.left, to_delete)
        node.right = self.dfs(node.right, to_delete)

        if node.val in to_delete:
            if node.left:
                self.forest.append(node.left)
            if node.right:
                self.forest.append(node.right)
            return None

        return node
```

872. Leaf-Similar Trees

```python
class Solution(object):
    def leafSimilar(self, root1, root2):
        """
        :type root1: Optional[TreeNode]
        :type root2: Optional[TreeNode]
        :rtype: bool
        """

        lvs_1 = self.dfs(root1)
        lvs_2 = self.dfs(root2)

        return lvs_1 == lvs_2

    def dfs(self, node):

        if not node.left and not node.right:
            return [node.val]

        left_, right_ = [], []
        if node.left:
            left_ = self.dfs(node.left)
        if node.right:
            right_ = self.dfs(node.right)

        return left_ + right_
```

1367. Linked List in Binary Tree

```python
class Solution(object):
    def isSubPath(self, head, root):
        """
        :type head: Optional[ListNode]
        :type root: Optional[TreeNode]
        :rtype: bool
        """

        if not root:
            return False

        return self.match(head, root) or self.isSubPath(head, root.left) or self.isSubPath(head, root.right)

    def match(self, head, node):

        if not head:
            return True
        if not node:
            return False
        if head.val != node.val:
            return False

        return self.match(head.next, node.left) or self.match(head.next, node.right)
```

701. Insert into a Binary Search Tree

```python
class Solution(object):
    def insertIntoBST(self, root, val):
        """
        :type root: Optional[TreeNode]
        :type val: int
        :rtype: Optional[TreeNode]
        """

        if not root:
            return TreeNode(val)

        self.dfs(root, val)

        return root

    def dfs(self, node, val):
        if not node:
            return TreeNode(val)

        if val < node.val:
            node.left = self.dfs(node.left, val)

        if val > node.val:
            node.right = self.dfs(node.right, val)

        return node
```

2331. Evaluate Boolean Binary Tree

```python
class Solution(object):
    def evaluateTree(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: bool
        """

        return self.dfs(root) == 1

    def dfs(self, node):
        if not node.left and not node.right:
            return node.val

        left = self.dfs(node.left)
        right = self.dfs(node.right)

        if node.val == 2:
            return left or right

        if node.val == 3:
            return left and right
```

654. Maximum Binary Tree

```python
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def constructMaximumBinaryTree(self, nums):
        """
        :type nums: List[int]
        :rtype: Optional[TreeNode]
        """

        return self.construct(nums)

    def construct(self, nums):

        if len(nums) == 0:
            return None

        max_ = float('-inf')
        pivot = None

        for i in range(len(nums)):
            if nums[i] > max_:
                max_ = nums[i]
                pivot = i

        node = TreeNode(max_)
        node.left = self.construct(nums[:pivot])
        node.right = self.construct(nums[pivot+1:])
        return node
```

814. Binary Tree Pruning

```python
class Solution(object):
    def pruneTree(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: Optional[TreeNode]
        """

        sum_ = self.dfs(root)

        if sum_ == 0:
            return None

        return root

    def dfs(self, node):
        if not node:
            return 0

        left_sum = self.dfs(node.left)
        right_sum = self.dfs(node.right)

        if left_sum == 0:
            node.left = None

        if right_sum == 0:
            node.right = None

        return node.val + left_sum + right_sum
```


1905. Count Sub Islands

```python
class Solution(object):
    def countSubIslands(self, grid1, grid2):
        """
        :type grid1: List[List[int]]
        :type grid2: List[List[int]]
        :rtype: int
        """

        if not grid2:
            return 0

        count = 0
        for i in range(len(grid2)):
            for j in range(len(grid2[0])):
                if grid2[i][j] == 1:
                    if self.dfs(grid1, grid2, i, j):
                        count += 1

        return count

    def dfs(self, grid1, grid2, i, j):
        # out of bounds or water in grid2
        if i<0 or j<0 or i>=len(grid2) or j>=len(grid2[0]) or grid2[i][j] == 0:
            return True

        grid2[i][j] = 0

        is_sub = (grid1[i][j] == 1)

        is_sub &= self.dfs(grid1, grid2, i+1, j)
        is_sub &= self.dfs(grid1, grid2, i-1, j)
        is_sub &= self.dfs(grid1, grid2, i, j+1)
        is_sub &= self.dfs(grid1, grid2, i, j-1)

        return is_sub
```

547. Number of Provinces

```python
class Solution(object):
    def findCircleNum(self, isConnected):
        """
        :type isConnected: List[List[int]]
        :rtype: int
        """

        n = len(isConnected)
        visited = [False]*n
        count = 0

        for i in range(n):
            if not visited[i]:
                self.dfs(isConnected, visited, i)
                count += 1

        return count

    def dfs(self, isConnected, visited, i):
        visited[i] = True

        for j in range(len(isConnected)):
            if isConnected[i][j] == 1 and visited[j] == False:
                self.dfs(isConnected, visited, j)
```

100. Same Tree

```python
class Solution(object):
    def isSameTree(self, p, q):
        """
        :type p: Optional[TreeNode]
        :type q: Optional[TreeNode]
        :rtype: bool
        """

        return self.dfs(p, q)

    def dfs(self, node_p, node_q):
        if not node_p and not node_q:
            return True
        elif not node_p or not node_q:
            return False
        elif node_p.val != node_q.val:
            return False
        return self.dfs(node_p.left, node_q.left) and self.dfs(node_p.right, node_q.right)
```

572. Subtree of Another Tree

```python
class Solution(object):
    def isSubtree(self, root, subRoot):
        """
        :type root: Optional[TreeNode]
        :type subRoot: Optional[TreeNode]
        :rtype: bool
        """

        if not root:
            return False

        if self.dfs(root, subRoot):
            return True

        return self.isSubtree(root.left, subRoot) or self.isSubtree(root.right, subRoot)

    def dfs(self, p, q):

        if not p and not q:
            return True

        if not p or not q:
            return False

        if p.val != q.val:
            return False

        # If the program comes to this point it means the p.val == q.val

        return self.dfs(p.left, q.left) and self.dfs(p.right, q.right)
```

114. Flatten Binary Tree to Linked List

```python
class Solution(object):
    def flatten(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: None Do not return anything, modify root in-place instead.
        """

        self.prev = None
        self.preorder(root)

    def preorder(self, node):

        if not node:
            return

        left = node.left
        right = node.right

        if self.prev:
            self.prev.left = None
            self.prev.right = node

        self.prev = node

        self.preorder(left)
        self.preorder(right)
```

1992. Find All Groups of Farmland

```python
class Solution(object):
    def findFarmland(self, land):
        """
        :type land: List[List[int]]
        :rtype: List[List[int]]
        """

        m = len(land)
        n = len(land[0])

        res = []

        for i in range(m):
            for j in range(n):
                if land[i][j] == 1:
                    self.i_min = float('inf')
                    self.i_max = float('-inf')
                    self.j_min = float('inf')
                    self.j_max = float('-inf')
                    self.dfs(land, i, j)
                    res.append([self.i_min, self.j_min, self.i_max, self.j_max])

        return res

    def dfs(self, land, i, j):

        if i<0 or j<0 or i>=len(land) or j>=len(land[0]) or land[i][j] != 1:
            return

        land[i][j] = 0

        # If i is smallest and j is smallest - record
        self.i_min = min(self.i_min, i)
        self.j_min = min(self.j_min, j)
        # If i is largest and j is largest - record
        self.i_max = max(self.i_max, i)
        self.j_max = max(self.j_max, j)

        self.dfs(land, i+1, j)
        self.dfs(land, i-1, j)
        self.dfs(land, i, j+1)
        self.dfs(land, i, j-1)

        return
```

108. Convert Sorted Array to Binary Search Tree

```python
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def sortedArrayToBST(self, nums):
        """
        :type nums: List[int]
        :rtype: Optional[TreeNode]
        """

        if len(nums) == 0:
            return None

        middle = len(nums) // 2
        node = TreeNode(nums[middle])

        node.left = self.sortedArrayToBST(nums[:middle])
        node.right = self.sortedArrayToBST(nums[middle+1:])

        return node
```

99. Recover Binary Search Tree

```python
class Solution(object):
    def recoverTree(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: None Do not return anything, modify root in-place instead.
        """

        self.first = None
        self.second = None
        self.prev = None
        self.inorder(root)
        self.first.val, self.second.val = self.second.val, self.first.val

    def inorder(self, node):
        if not node:
            return
        self.inorder(node.left)
        if self.prev and self.prev.val > node.val:
            if not self.first:
                self.first = self.prev
            self.second = node
        self.prev = node
        self.inorder(node.right)
```

530. Minimum Absolute Difference in BST

```python
class Solution(object):
    def getMinimumDifference(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: int
        """

        self.min = float('inf')
        self.prev = None
        self.inorder(root)
        return self.min

    def inorder(self, node):
        if not node:
            return
        self.inorder(node.left)
        if self.prev:
            self.min = min(self.min, abs(self.prev.val - node.val))
        self.prev = node
        self.inorder(node.right)
```

98. Validate Binary Search Tree

```python
class Solution(object):
    def isValidBST(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: bool
        """
        self.prev = None
        return self.inorder(root)
    def inorder(self, node):
        if not node:
            return True
        if not self.inorder(node.left):
            return False
        if self.prev and self.prev.val >= node.val:
            return False
        self.prev = node
        return self.inorder(node.right)
```

669. Trim a Binary Search Tree

```python
class Solution(object):
    def trimBST(self, root, low, high):
        """
        :type root: Optional[TreeNode]
        :type low: int
        :type high: int
        :rtype: Optional[TreeNode]
        """

        if not root:
            return None

        return self.dfs(root, low, high)

    def dfs(self, node, low, high):

        if not node:
            return None

        if node.val < low:
            return self.dfs(node.right, low, high)

        if node.val > high:
            return self.dfs(node.left, low, high)

        node.left = self.dfs(node.left, low, high)
        node.right = self.dfs(node.right, low, high)

        return node
```

96. Unique Binary Search Trees

```python
class Solution(object):
    def numTrees(self, n):
        """
        :type n: int
        :rtype: int
        """

        dp = [0]*(n+1)
        dp[0] = 1
        dp[1] = 1

        for node in range(2, n+1):
            for i in range(1, node+1):
                dp[node] += dp[i-1] * dp[node-i]

        return dp[n]
```
