##Simple Neural Network

```python
import numpy as np

class NeuralNetwork:
    def __init__(self, input_size, output_size, hidden_size):

        self.weights1 = np.random.randn(input_size, hidden_size)
        self.weights2 = np.random.randn(hidden_size, output_size)

        self.bias1 = np.random.randn(1, hidden_size)
        self.bias2 = np.random.randn(1, output_size)

    def relu(self, x):
        return np.maximum(0, x)

    def sigmoid(self, x):
        return 1/(1+np.exp(-x))

    def relu_derivative(self, x):
        return (x > 0).astype(float)

    def forward(self, X):
        self.z1 = X @ self.weights1 + self.bias1
        self.z2 = self.relu(self.z1)
        self.z3 = self.z2 @ self.weights2 + self.bias2
        self.output = self.sigmoid(self.z3)
        return self.output

    def backward(self, X, y, lr):

        d_output = self.output - y

        d_weights2 = self.z2.T @ d_output
        d_bias2 = np.sum(d_output, axis=0, keepdims=True)

        d_hidden = (d_output @ self.weights2.T) * self.relu_derivative(self.z1)

        d_weights1 = X.T @ d_hidden
        d_bias1 = np.sum(d_hidden, axis=0, keepdims=True)

        self.weights2 -= lr*d_weights2
        self.bias2 -= lr*d_bias2

        self.weights1 -= lr*d_weights1
        self.bias1 -= lr*d_bias1

    def train(self, X, y, epochs=1000, lr=0.01):

        for i in range(epochs):

            output = self.forward(X)
            self.backward(X, y, lr)

            if i % 100 == 0:
                loss = -np.mean(
                    y*np.log(output+1e-8) +
                    (1-y)*np.log(1-output+1e-8)
                )
                print("Loss:", loss)
```

##k-means clustering

```python
import numpy as np

class KMeans:

    def __init__(self, k, max_iters=100):
        self.k = k
        self.max_iters = max_iters


    def fit(self, X):

        n_samples = X.shape[0]

        # Step 1: randomly choose k points as initial centroids
        random_indices = np.random.choice(n_samples, self.k, replace=False)
        self.centroids = X[random_indices]

        for _ in range(self.max_iters):

            clusters = [[] for _ in range(self.k)]

            # Step 2: assign each point to the nearest centroid
            for i, point in enumerate(X):

                distances = []

                for centroid in self.centroids:
                    distance = np.linalg.norm(point - centroid)
                    distances.append(distance)

                closest_centroid = np.argmin(distances)
                clusters[closest_centroid].append(i)

            # Step 3: compute new centroids
            new_centroids = []

            for cluster in clusters:

                cluster_points = X[cluster]
                centroid = np.mean(cluster_points, axis=0)
                new_centroids.append(centroid)

            new_centroids = np.array(new_centroids)

            # Step 4: stop if centroids do not change
            if np.all(self.centroids == new_centroids):
                break

            self.centroids = new_centroids

        # store cluster labels
        self.labels = np.zeros(n_samples)

        for cluster_id, cluster in enumerate(clusters):
            for sample_index in cluster:
                self.labels[sample_index] = cluster_id


    def predict(self, X):

        labels = []

        for point in X:

            distances = []

            for centroid in self.centroids:
                distance = np.linalg.norm(point - centroid)
                distances.append(distance)

            labels.append(np.argmin(distances))

        return np.array(labels)
```

##Dynamic Programming

322. Coin Change

```python
class Solution(object):
    def coinChange(self, coins, amount):
        """
        :type coins: List[int]
        :type amount: int
        :rtype: int
        """

        dp = [amount + 1]*(amount + 1)
        dp[0] = 0

        for i in range(1, amount + 1):
            for c in coins:
                if i - c >= 0:
                    dp[i] = min(1 + dp[i-c], dp[i])

        return dp[-1] if dp[-1] != amount + 1 else -1
```

300. Longest Increasing Subsequence

```python
class Solution(object):
    def lengthOfLIS(self, nums):
        """
        :type nums: List[int]
        :rtype: int

        Time: O(len(nums)^2)
        Space: O(len(nums))
        """

        LIS = [1]*len(nums)

        for i in range(len(nums)-1, -1, -1):
            for j in range(i+1,len(nums)):
                if nums[i] < nums[j]:
                    LIS[i] = max(LIS[i], 1 + LIS[j])

        return max(LIS)
```

1143. Longest Common Subsequence

```python
class Solution(object):
    def longestCommonSubsequence(self, text1, text2):
        """
        :type text1: str
        :type text2: str
        :rtype: int

        Time: O(len(text1) * len(text2))
        Space: O(len(text1) * len(text2))
        """

        dp = [[0 for i in range(len(text2)+1)] for j in range(len(text1)+1)]

        for i in range(len(text1)-1, -1, -1):
            for j in range(len(text2)-1, -1, -1):
                if text1[i] == text2[j]:
                    dp[i][j] = 1 + dp[i+1][j+1]
                else:
                    dp[i][j] = max(dp[i+1][j], dp[i][j+1])

        return dp[0][0]
```

198. House Robber

```python
class Solution(object):
    def rob(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """

        rob1, rob2 = 0, 0

        for n in nums:
            temp = max(n + rob2, rob1)
            rob2 = rob1
            rob1 = temp

        return temp
```

213. House Robber II

```python
class Solution(object):
    def rob(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """

        if len(nums) == 1:
            return nums[0]

        def helper(nums):
            rob1, rob2 = 0, 0

            for n in nums:
                temp = max(n + rob2, rob1)
                rob2 = rob1
                rob1 = temp

            return temp

        return max(helper(nums[1:]), helper(nums[:-1]))
```
